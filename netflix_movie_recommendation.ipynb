```python
import numpy as np
import pandas as pd
import pickle as pk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv("Netflix_TV_Shows_and_Movies.csv")
data
```

```python
kids = ["PG", "TV-PG", "G", "TV-Y7", "TV-Y", "TV-G"]
adult = ["TV-MA", "R", "TV-14", "PG-13", "NC-17"]
data["age_certification"] = data["age_certification"].replace(
    {i : "childish family" for i in kids} |
    {i : "adult" for i in adult}
    )

for i in data.columns:
    data[i] = data[i].fillna(" ")
```

```python
data_text = (data["title"] + " " + data["type"] + " " + str(data["release_year"]) + " " + data["age_certification"])
data_text = data_text.str.lower()
type(data_text)
```

```python
model = TfidfVectorizer()

data_number = model.fit_transform(data_text)
```

```python
x_text = input("Describe what you're in the mood for, and I'll suggest the best matches : ").lower()
x_number = model.transform([x_text])
x_text
```

```python
similarity = cosine_similarity(x_number,data_number).flatten()
top_3 = np.argsort(-similarity)[:3]
counter = 1
for i in top_3:
    print(counter)
    print("اسم فیلم :",data["title"][i])
    print("داستان فیلم :",data["description"][i])
    print("➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖")
    counter += 1
```

```python
with open("model.pkl", "wb")as file:
    pk.dump(model, file)

data.to_pickle("Netflix_TV_Shows_and_Movies.pkl")
```
